# 26. 저장성 축으로 갈라 출하량 효과를 다시 재기

## Overview

21번과 25번이 같은 질문에 다른 답을 냈습니다.

| 노트북 | 대상 | 커버리지 40% 효과 |
|---|---|---|
| 21 | 출하량 있던 6품목 | **+2.97%p** |
| 25 | 중간지대 4품목 | **+0.22%p** |

25번 결론에서 이유를 짐작했습니다: **21번의 6품목에는 양파(165일)·감자(45일)·
생강(90일) 같은 저장성 품목이 있었고, 25번의 중간지대는 셋이 엽채류**
(상추 10일·얼갈이배추 10일·풋고추 14일)였습니다.

**짐작을 숫자로 바꿉니다.** 출하량이 있는 9품목을 저장일수로 갈라 같은
실험을 반복합니다.

## 왜 저장성이 갈림길일 수 있나

**저장이 되면 재고가 완충 역할을 합니다.** 오늘 많이 들어와도 창고로 가서
가격이 천천히 움직이고, 그 시차가 **예측할 거리**를 만듭니다.

**저장이 안 되면 그날 물량을 그날 팝니다.** 출하량이 즉시 가격에 반영되므로,
가격 자체(모멘텀·계절 편차)가 이미 그 정보를 담고 있습니다. 출하량을 따로
넣어도 새로 알려주는 게 없습니다.

이 짐작이 맞다면 **장기 저장군에서만 효과가 나야** 합니다.

## 묶음

14번의 저장일수를 그대로 씁니다(확인·추정 구분 포함).

| 묶음 | 품목 | 저장일수 |
|---|---|---|
| **장기 (45일 이상)** | 양파 · 생강 · 감자 | 165 · 90 · 45 |
| **단기 (30일 이하)** | 배추 · 풋고추 · 상추 · 얼갈이배추 · 미나리 · 파 | 30 · 14 · 10 · 10 · 7 · 7 |

**생강이 걸립니다.** 19번에서 정체 품목(정체율 58.2%)으로 판정돼 운영 대상에서
뺐습니다. 가격이 자주 멈추는 품목이라 방향이라는 개념 자체가 흐릿합니다.
그래서 **생강을 뺀 2품목(양파·감자)으로도 같이 재서** 판정이 생강 때문에
생긴 게 아닌지 확인합니다.

## 가설

| 가설 | 내용 | 맞다면 보여야 할 것 |
|---|---|---|
| **H1** | 장기 저장군에서 출하량이 값을 한다 | 노이즈를 넘는 상승, 특히 운영 지점(30~40%) |
| **H2** | 단기 저장군에서는 값을 안 한다 | 전 구간 노이즈 이내 |
| **H3** | 판정이 생강 때문이 아니다 | 양파·감자만으로도 H1 이 유지됨 |

**H1 이 틀릴 수 있는 이유**: 장기 저장군이 3품목뿐이라 노이즈 플로어가 넓어집니다
(교훈 7). 효과가 있어도 못 잡을 수 있고, 없는데 잡을 수도 있습니다.

**어느 쪽이든 정보입니다.** 갈리면 "저장성 있는 품목에만 출하량을 붙인다"는
수집 방침이 서고, 안 갈리면 21번의 +2.97%p 를 다른 이유로 설명해야 합니다.

## 1. 준비

15·21·25번과 같은 피처·폴드·하이퍼파라미터입니다. **학습은 19품목 전부**
(19번에서 확정), 평가만 묶음별로 바꿉니다.

```python
import time
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from matplotlib import font_manager
from sklearn.ensemble import HistGradientBoostingClassifier

_av = {f.name for f in font_manager.fontManager.ttflist}
for _f in ("AppleGothic", "Apple SD Gothic Neo", "NanumGothic", "Malgun Gothic"):
    if _f in _av:
        plt.rcParams["font.family"] = _f
        break
sns.set_style("whitegrid", {"font.family": plt.rcParams["font.family"]})
plt.rcParams["axes.unicode_minus"] = False

ROOT = Path.cwd()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent
df = pd.read_parquet(ROOT / "notebooks" / ".cache" / "mart.parquet")
df["price_date"] = pd.to_datetime(df["price_date"])
df = df.sort_values(["item", "price_date"]).reset_index(drop=True)
all_items = sorted(df.item.unique())

# 14번의 저장일수
STORAGE = {
    "시금치": 7, "상추": 10, "배추": 30, "양배추": 50, "무": 45, "감자": 45,
    "양파": 165, "얼갈이배추": 10, "알배기배추": 20, "깻잎": 7, "미나리": 7,
    "열무": 7, "파": 7, "풋고추": 14, "브로콜리": 14, "고구마": 60,
    "생강": 90, "깐마늘(국산)": 30, "피마늘": 180,
}
SHIP = ["shipment_qty", "shipment_amt", "shipment_qty_w1ago", "shipment_qty_w2ago",
        "shipment_qty_w3ago", "shipment_qty_w4ago", "shipment_market_cnt"]

# 출하량이 실제로 있는 품목만 대상
채움 = df.groupby("item").shipment_qty.apply(lambda s: s.notna().mean() * 100)
있는품목 = sorted(채움.index[채움 > 50])
장기 = [i for i in 있는품목 if STORAGE[i] >= 45]
단기 = [i for i in 있는품목 if STORAGE[i] <= 30]
장기_생강뺀 = [i for i in 장기 if i != "생강"]

표 = pd.DataFrame({
    "저장일수": {i: STORAGE[i] for i in 있는품목},
    "출하량 채움 %": 채움[있는품목].round(1),
    "묶음": {i: ("장기" if STORAGE[i] >= 45 else "단기") for i in 있는품목},
}).sort_values("저장일수", ascending=False)
display(표)
print(f"장기 {len(장기)}품목: {장기}")
print(f"단기 {len(단기)}품목: {단기}")
print(f"장기(생강 제외) {len(장기_생강뺀)}품목: {장기_생강뺀}")
```

```python
H = 7
EMBARGO = pd.Timedelta(days=int(H * 1.5))
FOLDS = [("2023-01-01", "2023-06-30"), ("2023-07-01", "2023-12-31"),
         ("2024-01-01", "2024-06-30"), ("2024-07-01", "2024-12-31"),
         ("2025-01-01", "2025-06-30"), ("2025-07-01", "2025-12-31"),
         ("2026-01-01", "2026-12-31")]
PARAMS = dict(learning_rate=0.014, max_iter=502, min_samples_leaf=21,
              max_leaf_nodes=15, l2_regularization=1.0)
CAL = ["month", "dayofweek", "days_to_major_holiday", "days_since_major_holiday",
       "days_since_solar_term"]
COVS = [0.8, 0.6, 0.5, 0.4, 0.3, 0.2]     # 10% 는 25번에서 표본이 얇아 뺀다

b = df.copy()
b["_logp"] = np.log(b.price_kg_avg)


def g(c):
    return b.groupby("item")[c]


b["y"] = g("_logp").shift(-H) - b._logp
b["오름"] = b.y > 0
b["미래가격"] = g("price_kg_avg").shift(-H)
for lag in (1, 2, 3, 5, 10, 20):
    b[f"ret{lag}"] = b._logp - g("_logp").shift(lag)
for w in (5, 20, 60):
    mp = max(2, w // 3)
    b[f"dev{w}"] = b._logp - g("_logp").transform(
        lambda s: s.rolling(w, min_periods=mp).mean())
    b[f"vol{w}"] = g("_logp").transform(
        lambda s: s.diff().rolling(w, min_periods=mp).std())
doy = b.price_date.dt.dayofyear
b["doy_sin"] = np.sin(2 * np.pi * doy / 365.25)
b["doy_cos"] = np.cos(2 * np.pi * doy / 365.25)
b["seas_dev"] = b._logp - (b.groupby(["item", "month"])["_logp"]
                           .transform(lambda s: s.shift(1).expanding(min_periods=20).mean()))
b["item_code"] = pd.Categorical(b.item, categories=all_items).codes

b["_logq"] = np.log1p(b.shipment_qty)
b["ship_ret1"] = b._logq - g("_logq").shift(1)
b["ship_dev20"] = b._logq - g("_logq").transform(
    lambda s: s.rolling(20, min_periods=7).mean())
b["ship_price"] = b.shipment_amt / b.shipment_qty.replace(0, np.nan)
SHIP_F = SHIP + ["ship_ret1", "ship_dev20", "ship_price"]

feat = b.dropna(subset=["y", "미래가격"]).reset_index(drop=True)
BASE = ([f"ret{l}" for l in (1, 2, 3, 5, 10, 20)] + [f"dev{w}" for w in (5, 20, 60)]
        + [f"vol{w}" for w in (5, 20, 60)] + CAL
        + ["doy_sin", "doy_cos", "seas_dev", "item_code"])
print(f"학습·평가 대상 {len(feat):,}행 | 기본 {len(BASE)}피처 · +출하량 {len(BASE+SHIP_F)}피처")


def 돌리기(use, 평가품목, noise=None, seed=None):
    d, use = feat, list(use)
    if noise:
        d = feat.copy()
        rng = np.random.default_rng(seed)
        for j in range(noise):
            d[f"noise{j}"] = rng.normal(size=len(d))
            use.append(f"noise{j}")
    cat = [use.index(c) for c in ("item_code", "month", "dayofweek")]
    out = []
    for s, e in FOLDS:
        s, e = pd.Timestamp(s), pd.Timestamp(e)
        tr = d[d.price_date < s - EMBARGO]
        te = d[(d.price_date >= s) & (d.price_date <= e) & (d.item.isin(평가품목))]
        clf = HistGradientBoostingClassifier(
            categorical_features=sorted(cat), early_stopping=False,
            random_state=42, **PARAMS).fit(tr[use], tr.오름)
        t = te[["item", "price_date", "오름", "price_kg_avg", "미래가격"]].copy()
        t["오를확률"] = clf.predict_proba(te[use])[:, 1]
        out.append(t)
    r = pd.concat(out).reset_index(drop=True)
    r["방향맞음"] = (r.오를확률 > 0.5) == r.오름
    r["확신도"] = (r.오를확률 - 0.5).abs()
    return r


def 컷(r, cov):
    q = r.확신도.quantile(1 - cov)
    s = r[r.확신도 >= q]
    넣음 = (r.오를확률 > 0.5) & (r.확신도 >= q)
    실현 = np.where(넣음, r.미래가격.values, r.price_kg_avg.values)
    더받은돈 = (실현.mean() / r.price_kg_avg.mean() - 1) * 100
    비율 = 넣음.mean()
    return {"적중률 %": s.방향맞음.mean() * 100, "남는 날": len(s),
            "더 받은 돈 %": 더받은돈,
            "창고비 한도 %": 더받은돈 / 비율 if 비율 > 0 else np.nan}


def 묶음비교(품목, 이름):
    # 한 묶음에서 기본 vs 기본+출하량을 같은 커버리지로 비교한다.
    r_a = 돌리기(BASE, 품목)
    r_s = 돌리기(BASE + SHIP_F, 품목)
    nz = []
    for seed in (0, 1, 2, 3, 4):
        rn = 돌리기(BASE, 품목, noise=4, seed=seed)
        for cov in COVS:
            nz.append({"커버리지": cov, **컷(rn, cov)})
    nz = pd.DataFrame(nz)
    fl = nz.groupby("커버리지")["적중률 %"].agg(lambda s: s.max() - s.min())
    flh = nz.groupby("커버리지")["창고비 한도 %"].agg(lambda s: s.max() - s.min())
    rows = []
    for cov in COVS:
        a, s_ = 컷(r_a, cov), 컷(r_s, cov)
        rows.append({
            "묶음": 이름, "품목수": len(품목), "커버리지 %": cov * 100,
            "기본 적중률 %": a["적중률 %"], "출하량 적중률 %": s_["적중률 %"],
            "차이 %p": s_["적중률 %"] - a["적중률 %"], "노이즈 %p": fl[cov],
            "남는 날": a["남는 날"],
            "한도 차이 %p": s_["창고비 한도 %"] - a["창고비 한도 %"],
            "한도 노이즈 %p": flh[cov],
        })
    out = pd.DataFrame(rows)
    out["넘김"] = out["차이 %p"].abs() > out["노이즈 %p"]
    return out
```

## 2. 묶음별로 커버리지를 맞춰 비교

21·25번과 같은 방법입니다: 임계값이 아니라 **분위수**로 잘라 두 구성이 같은
수의 날을 남기게 합니다. 노이즈 플로어는 **묶음마다 따로** 잽니다(교훈 7:
대상이 바뀌면 다시 재야 합니다).

```python
t0 = time.time()
결과 = pd.concat([묶음비교(장기, "장기 (45일+)"),
                 묶음비교(단기, "단기 (30일-)")], ignore_index=True)
display(결과.round(2))
print(f"({time.time()-t0:.0f}초)")
print("\n21번(혼합 6품목) 은 커버리지 40% 에서 +2.97%p, 25번(중간지대 4품목) 은 +0.22%p 였습니다.")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 5.4))

for ax, (이름, 색) in zip(axes, [("장기 (45일+)", "tab:green"),
                                 ("단기 (30일-)", "tab:blue")]):
    s = 결과[결과.묶음 == 이름]
    x = np.arange(len(s))
    cols = [색 if v else "lightgray" for v in s["넘김"]]
    ax.bar(x, s["차이 %p"], color=cols)
    for i, (v, n) in enumerate(zip(s["차이 %p"], s["노이즈 %p"])):
        ax.plot([i - 0.4, i + 0.4], [n, n], color="tab:orange", lw=1.6)
        ax.plot([i - 0.4, i + 0.4], [-n, -n], color="tab:orange", lw=1.6)
        ax.text(i, v + (0.2 if v >= 0 else -0.5), f"{v:+.2f}", ha="center",
                fontsize=9)
    ax.axhline(0, color="black", lw=1.2)
    ax.axhline(2.97, color="tab:red", ls="--", lw=1.2, label="21번 +2.97%p")
    ax.set_xticks(x)
    ax.set_xticklabels([f"{v:.0f}%" for v in s["커버리지 %"]], fontsize=9)
    ax.set_xlabel("커버리지")
    ax.set_ylabel("출하량 효과 (%p)")
    ax.set_title(f"{이름}: {s.품목수.iloc[0]}품목", fontsize=11)
    ax.legend(fontsize=8.5)

ymin = min(결과["차이 %p"].min() - 1, -1)
ymax = max(결과["차이 %p"].max() + 1, 3.5)
for ax in axes:
    ax.set_ylim(ymin, ymax)
plt.tight_layout()
plt.show()
print("주황 선(노이즈)을 넘어야 색이 칠해집니다. 회색은 '차이 없음' 입니다.")
```

## 3. 생강을 빼면

생강은 19번에서 **정체 품목**(정체율 58.2%)으로 판정돼 운영 대상에서 뺐습니다.
가격이 자주 멈추는 품목이라 방향이라는 개념이 흐릿한데, 장기 저장군 3개 중
하나입니다.

**판정이 생강 때문에 생긴 게 아닌지** 확인합니다.

```python
t0 = time.time()
생강뺀결과 = 묶음비교(장기_생강뺀, "장기, 생강 제외 (양파·감자)")
비교 = pd.concat([결과[결과.묶음 == "장기 (45일+)"], 생강뺀결과], ignore_index=True)
display(비교[["묶음", "품목수", "커버리지 %", "기본 적중률 %", "출하량 적중률 %",
              "차이 %p", "노이즈 %p", "남는 날", "넘김"]].round(2))
print(f"({time.time()-t0:.0f}초)")
```

```python
fig, ax = plt.subplots(figsize=(10, 5.4))
for 이름, 색, mk in [("장기 (45일+)", "tab:green", "o"),
                    ("장기, 생강 제외 (양파·감자)", "tab:olive", "s")]:
    s = 비교[비교.묶음 == 이름]
    ax.plot(s["커버리지 %"], s["차이 %p"], f"{mk}-", ms=7, color=색,
            label=f"{이름} ({s.품목수.iloc[0]}품목)")
    ax.fill_between(s["커버리지 %"], -s["노이즈 %p"] / 2, s["노이즈 %p"] / 2,
                    color=색, alpha=0.12)
ax.axhline(0, color="black", lw=1.2)
ax.invert_xaxis()
ax.set_xlabel("커버리지 %  <- 오른쪽일수록 많이 잘랐음")
ax.set_ylabel("출하량 효과 (%p)")
ax.set_title("생강을 빼도 판정이 유지되는가 (옅은 띠가 각자의 노이즈 구간)",
             fontsize=11)
ax.legend(fontsize=9)
plt.tight_layout()
plt.show()
```

## 4. 품목별 저장일수와 효과

묶음이 아니라 **품목 낱개**로도 봅니다. 저장일수가 길수록 효과가 큰지
직접 확인합니다.

**⚠️ 품목 단위 노이즈는 1.3~2.8%p 로 넓습니다**(16·17번). 낱개 값은 상한으로
읽고, 추세만 봅니다.

```python
t0 = time.time()
rows = []
for it in 있는품목:
    a = 돌리기(BASE, [it])
    s_ = 돌리기(BASE + SHIP_F, [it])
    qa = a.확신도.quantile(1 - 0.4)
    qs = s_.확신도.quantile(1 - 0.4)
    rows.append({
        "item": it, "저장일수": STORAGE[it],
        "기본 적중률 %": a[a.확신도 >= qa].방향맞음.mean() * 100,
        "출하량 적중률 %": s_[s_.확신도 >= qs].방향맞음.mean() * 100,
        "정체 품목": it in ("생강",),
    })
낱개 = pd.DataFrame(rows).set_index("item")
낱개["차이 %p"] = (낱개["출하량 적중률 %"] - 낱개["기본 적중률 %"])
낱개 = 낱개.sort_values("저장일수", ascending=False)
display(낱개.round(2))

sp = 낱개.저장일수.corr(낱개["차이 %p"], method="spearman")
sp2 = (낱개[~낱개["정체 품목"]].저장일수
       .corr(낱개[~낱개["정체 품목"]]["차이 %p"], method="spearman"))
print(f"\n저장일수와 출하량 효과의 Spearman 상관: {sp:+.3f}")
print(f"  생강 제외: {sp2:+.3f}")
print(f"({time.time()-t0:.0f}초)")
print("양수면 '저장이 길수록 출하량이 값을 한다' 입니다. 9품목뿐이라 참고용입니다.")
```

```python
fig, ax = plt.subplots(figsize=(9.5, 5.4))
cols = ["tab:red" if v else "tab:blue" for v in 낱개["정체 품목"]]
ax.scatter(낱개.저장일수, 낱개["차이 %p"], s=90, c=cols, alpha=0.85)
for it, row in 낱개.iterrows():
    ax.annotate(it, (row.저장일수, row["차이 %p"]), fontsize=9,
                xytext=(6, 4), textcoords="offset points")
ax.axhline(0, color="black", lw=1.2)
ax.axvline(45, color="gray", ls="--", lw=1.2)
ax.text(46, 낱개["차이 %p"].max(), "45일: 장기/단기 경계", fontsize=8.5,
        color="gray")
ax.set_xscale("log")
ax.set_xlabel("저장 가능 일수 (로그 눈금)")
ax.set_ylabel("출하량 효과 (%p, 커버리지 40%)")
ax.set_title(f"오른쪽 위로 갈수록 짐작대로입니다 (상관 {sp:+.3f}, 빨강=정체 품목)",
             fontsize=11)
plt.tight_layout()
plt.show()
```

## 5. 정리

### 한 줄 요약

**짐작이 정반대였습니다.** 출하량은 저장성 **긴** 품목이 아니라 **짧은** 품목에서
값을 합니다. 저장일수와 효과의 상관이 **-0.723** 입니다.

그리고 더 정확히는 저장성 축이 아니라 **품목 둘**(파·미나리)입니다.

### 가설 판정

| 가설 | 내용 | 결과 | 판정 |
|---|---|---|---|
| **H1** | 장기 저장군에서 값을 한다 | 6개 커버리지 중 2개만 노이즈 넘김, **운영 지점(40%)은 -0.19%p** | **반증됨** |
| **H2** | 단기 저장군에서는 값을 안 한다 | **6개 커버리지 전부** 노이즈 넘김 | **반증됨** |
| **H3** | 판정이 생강 때문이 아니다 | 생강 빼도 장기는 여전히 약함 | **확인됨** |

| 묶음 | 커버리지 40% | 노이즈 | 넘김 구간 |
|---|---|---|---|
| 장기 3품목 (45일+) | **-0.19%p** | 0.96 | 2 / 6 |
| **단기 6품목 (30일-)** | **+1.87%p** | 0.72 | **6 / 6** |

**정확히 거꾸로입니다.** 25번 결론에 적었던 "저장이 되면 재고가 완충 역할을
해서 시차가 생기고, 그 시차가 예측할 거리를 만든다"는 짐작이 데이터와
어긋납니다.

### 진짜 원인은 품목 둘입니다
